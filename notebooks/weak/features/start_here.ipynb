Weak Features
=============

These features build on the topic’s main example. Each introduction explains the method and its trade-offs; the worked scripts retain their model choices and detailed calculations.

__Choose a Feature__

- [Strong Lensing](https://github.com/PyAutoLabs/autolens_workspace/blob/main/notebooks/weak/features/strong_lensing/start_here.ipynb): Strong and weak lensing probe different parts of a mass distribution. A combined analysis connects central strongly lensed features with the wider weak-lensing field through a shared mass model.

For the common setup, begin with [weak/start_here.ipynb](https://github.com/PyAutoLabs/autolens_workspace/blob/main/notebooks/weak/start_here.ipynb). Run Python scripts from the workspace root.
